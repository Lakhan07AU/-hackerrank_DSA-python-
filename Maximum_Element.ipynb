```python
def getMax(operations):
    stack = []
    max_stack = []
    result = []

    for operation in operations:
        op = operation.split()

        if op[0] == '1':  # Push operation
            x = int(op[1])
            stack.append(x)
            # Maintain the max_stack with the maximum element
            if not max_stack or x >= max_stack[-1]:
                max_stack.append(x)
            else:
                max_stack.append(max_stack[-1])

        elif op[0] == '2':  # Pop operation
            if stack:
                stack.pop()
                max_stack.pop()

        elif op[0] == '3':  # Max operation
            if max_stack:
                result.append(max_stack[-1])

    return result

# Input reading
n = int(input())  # Read number of operations
operations = [input().strip() for _ in range(n)]

# Get the results for each type 3 query
result = getMax(operations)

# Output the results for type 3 queries
for res in result:
    print(res)
print(getMax(result))
```
