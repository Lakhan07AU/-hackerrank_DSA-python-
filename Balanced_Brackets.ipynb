```python
def isBalanced(s):
    stack = []
    match = {')': '(', '}': '{', ']': '['}

    for char in s:
        if char in match.values():
            stack.append(char)
        elif char in match:
            if not stack or stack.pop() != match[char]:
                return "NO"

    return "YES" if not stack else "NO"
n = int(input())
for _ in range(n):
    s = input().strip()
    print(isBalanced(s))
```
