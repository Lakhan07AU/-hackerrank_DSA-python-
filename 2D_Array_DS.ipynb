```python
def hourglassSum(arr):
    max_sum = -float('inf')  # Initialize to a very small number to handle negative sums

    # Traverse each potential hourglass starting point
    for i in range(4):  # We can only start an hourglass at rows 0 to 3
        for j in range(4):  # We can only start an hourglass at columns 0 to 3
            # Calculate the sum of the current hourglass
            top = sum(arr[i][j:j+3])  # Sum of top row (3 elements)
            middle = arr[i+1][j+1]   # Middle element (1 element)
            bottom = sum(arr[i+2][j:j+3])  # Sum of bottom row (3 elements)

            hourglass_sum = top + middle + bottom

            # Update the maximum hourglass sum
            max_sum = max(max_sum, hourglass_sum)

    return max_sum

# Input reading
arr = [list(map(int, input().split())) for _ in range(6)]

# Output the result of the maximum hourglass sum
print(hourglassSum(arr))
```
