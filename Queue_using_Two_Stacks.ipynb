```python
class QueueUsingTwoStacks:
    def __init__(self):
        self.stack1 = []  # Input stack
        self.stack2 = []  # Output stack

    def enqueue(self, x):
        # Push element onto stack1
        self.stack1.append(x)

    def dequeue(self):
        # If stack2 is empty, move elements from stack1 to stack2
        if not self.stack2:
            while self.stack1:
                self.stack2.append(self.stack1.pop())
        # Pop the top of stack2 (which is the front of the queue)
        if self.stack2:
            return self.stack2.pop()
        return None  # In case of an empty queue

    def front(self):
        # If stack2 is empty, move elements from stack1 to stack2
        if not self.stack2:
            while self.stack1:
                self.stack2.append(self.stack1.pop())
        # Return the top element of stack2 (which is the front of the queue)
        if self.stack2:
            return self.stack2[-1]
        return None  # In case of an empty queue


# Read input
q = int(input())  # Number of queries
queue = QueueUsingTwoStacks()

for _ in range(q):
    query = list(map(int, input().split()))
    if query[0] == 1:
        # Enqueue operation
        queue.enqueue(query[1])
    elif query[0] == 2:
        # Dequeue operation
        queue.dequeue()
    elif query[0] == 3:
        # Print front operation
        print(queue.front())
```
