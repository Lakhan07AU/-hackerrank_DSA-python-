```python
def equalStacks(h1, h2, h3):
    # Calculate the cumulative heights for each stack
    sum1, sum2, sum3 = sum(h1), sum(h2), sum(h3)

    # Start with pointers at the beginning of each stack
    i, j, k = 0, 0, 0

    # Keep reducing the stack heights until they are equal
    while sum1 != sum2 or sum2 != sum3:
        # Find the stack with the greatest height and remove the top cylinder
        if sum1 > sum2 or sum1 > sum3:
            sum1 -= h1[i]
            i += 1
        elif sum2 > sum1 or sum2 > sum3:
            sum2 -= h2[j]
            j += 1
        else:
            sum3 -= h3[k]
            k += 1

    return sum1  # All sums are now equal

# Input reading
n1, n2, n3 = map(int, input().split())
h1 = list(map(int, input().split()))
h2 = list(map(int, input().split()))
h3 = list(map(int, input().split()))

# Output the result of equalizing stack heights
print(equalStacks(h1, h2, h3))
```
